# Model Comparison — Competition-Aware CV

**Scoring**: `Score = TP×10 − FP×5 − NoVariables×200`  (cap: 1 000 contacts per submission)

**CV setup**: 5-fold StratifiedKFold → each validation fold is exactly **1 000 observations**.  
For each fold the model can contact *up to* 1 000 customers; `sweep_cutoff` finds the optimal N ≤ 1 000 that maximises the score. Contacting fewer is valid and sometimes better.

**Baseline to beat**: **3 885** (LR-L1, 5 features, from `diagnostics_mp.py`)

```python
import sys, warnings, copy, time
from pathlib import Path
warnings.filterwarnings('ignore')

sys.path.insert(0, str(Path('..') / 'src'))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns

from data import load_data
from features import l1_path_ranking, mutual_info_ranking
from models import get_models
from cv import competition_cv_evaluate
from scoring import sweep_cutoff
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, cross_val_predict

FIGURES = Path('figures')
FIGURES.mkdir(exist_ok=True)
BASELINE = 3885

sns.set_theme(style='whitegrid', palette='muted', font_scale=1.1)
plt.rcParams['figure.dpi'] = 120
plt.rcParams['savefig.bbox'] = 'tight'

X_train, y_train, X_test, Xs_train, Xs_test, scaler = load_data()
print(f'Train: {X_train.shape}   Test: {X_test.shape}')
print(f'Positive rate: {y_train.mean():.3f}')
```

## 1. Feature Selection: L1 Path vs Mutual Information

We compare two unsupervised-style feature importance signals:
- **L1 path order**: features that enter the Lasso path at lower regularization are more predictive
- **Mutual information**: measures non-linear dependence between feature and label

Agreement between methods strengthens confidence in a feature; disagreement warrants closer inspection.

```python
print('Computing L1 path ranking...')
l1_rank = l1_path_ranking(Xs_train, y_train)

print('Computing mutual information ranking...')
mi_rank = mutual_info_ranking(X_train, y_train)

TOP_K = 15
l1_top = set(l1_rank[:TOP_K].tolist())
mi_top = set(mi_rank[:TOP_K].tolist())
overlap = l1_top & mi_top

df_feat = pd.DataFrame({
    'L1 rank': {j: i+1 for i, j in enumerate(l1_rank[:TOP_K])},
    'MI rank': {j: list(mi_rank).index(j)+1 if j in mi_rank[:TOP_K] else '>15' for j in l1_top},
}).fillna('>15')
df_feat.index = [f'V{j+1}' for j in df_feat.index]
df_feat['In both top-15'] = df_feat.index.map(lambda x: '✓' if int(x[1:])-1 in overlap else '')

print(f'\nTop-{TOP_K} features by each method (overlap = {len(overlap)}/{TOP_K}):')
display(df_feat.sort_values('L1 rank'))
```

```python
# Side-by-side bar of top-10 L1 vs MI ranks
k_show = 10
l1_labels = [f'V{j+1}' for j in l1_rank[:k_show]]
mi_labels  = [f'V{j+1}' for j in mi_rank[:k_show]]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
colors_l1 = ['#e15759' if f in [f'V{j+1}' for j in overlap] else '#4e79a7' for f in l1_labels]
colors_mi = ['#e15759' if f in [f'V{j+1}' for j in overlap] else '#f28e2b' for f in mi_labels]

axes[0].barh(l1_labels[::-1], range(k_show, 0, -1), color=colors_l1[::-1], edgecolor='white')
axes[0].set_title('L1 Path: Top-10 Features')
axes[0].set_xlabel('Rank importance (higher = more important)')

axes[1].barh(mi_labels[::-1], range(k_show, 0, -1), color=colors_mi[::-1], edgecolor='white')
axes[1].set_title('Mutual Information: Top-10 Features')
axes[1].set_xlabel('Rank importance')

red_patch  = mpatches.Patch(color='#e15759', label='In both top-15')
blue_patch = mpatches.Patch(color='#4e79a7', label='L1 only')
org_patch  = mpatches.Patch(color='#f28e2b', label='MI only')
fig.legend(handles=[red_patch, blue_patch, org_patch], loc='lower center',
           ncol=3, bbox_to_anchor=(0.5, -0.08))

plt.tight_layout()
plt.savefig(FIGURES / '07_feat_selection.png')
plt.show()
```

## 2. Score vs Number of Features (LR-L1 sweep)

Each additional feature costs EUR 200 (NoVariables × 200). This plot shows the trade-off: adding a feature improves prediction but also increases the fixed cost. The optimal k is where the marginal benefit of an extra feature drops below 200.

```python
skf5 = StratifiedKFold(5, shuffle=True, random_state=0)
k_range = range(1, 31)
k_scores = []

print('Sweeping k=1..30 features for LR-L1...')
for k in k_range:
    feats = l1_rank[:k]
    p_cv = cross_val_predict(
        LogisticRegression(max_iter=2000, random_state=0),
        Xs_train[:, feats], y_train, cv=skf5, method='predict_proba'
    )[:, 1]
    # aggregate: split into 5 folds manually to get per-fold scores
    fold_scores = []
    for _, val_idx in skf5.split(Xs_train, y_train):
        s, *_ = sweep_cutoff(y_train[val_idx], p_cv[val_idx], k)
        fold_scores.append(s)
    k_scores.append(np.mean(fold_scores))

best_k = int(np.argmax(k_scores)) + 1
print(f'Best k: {best_k}  (mean CV score = {k_scores[best_k-1]:.0f})')
```

```python
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(list(k_range), k_scores, 'o-', color='#4e79a7', linewidth=2, markersize=5)
ax.axhline(BASELINE, color='#e15759', linestyle='--', linewidth=1.5, label=f'Baseline = {BASELINE}')
ax.axvline(best_k, color='#59a14f', linestyle=':', linewidth=1.5, label=f'Best k = {best_k}')
ax.fill_between(list(k_range), k_scores, BASELINE,
                where=[s < BASELINE for s in k_scores],
                alpha=0.15, color='#e15759', label='Below baseline')
ax.set_xlabel('k (number of features used)')
ax.set_ylabel('Mean CV competition score')
ax.set_title('Score vs Features Used (LR-L1, 5-fold CV)')
ax.legend()
plt.tight_layout()
plt.savefig(FIGURES / '08_score_vs_k.png')
plt.show()
```

## 3. Model Comparison: CV Competition Scores

We evaluate five model families using the same 5-fold competition-aware CV loop. For tree-based models, `n_vars_used` counts only features with non-zero importance — the NoVariables cost reflects actual feature usage, not the full 500.

**Models compared**:
- `lr_l1`: Logistic Regression with L1 (Lasso) penalty — explicit sparsity, low NoVariables cost
- `lr_l2`: Logistic Regression with L2 (Ridge) penalty — all 500 features, high NoVariables cost
- `rf`: Random Forest — ensemble, selects features implicitly via splits
- `lgbm` / `gbm`: Gradient Boosting — often highest accuracy, but variable NoVariables cost
- `xgb`: XGBoost — similar to LGBM, different regularization

```python
models = get_models()
results = {}

for name, model in models.items():
    print(f'Evaluating {name}...', end=' ', flush=True)
    t0 = time.time()
    results[name] = competition_cv_evaluate(model, Xs_train, y_train)
    print(f'{time.time()-t0:.1f}s  '
          f'mean_score={results[name]["mean_score"]:.0f} ± {results[name]["std_score"]:.0f}  '
          f'mean_N={results[name]["mean_N"]:.0f}  mean_nvars={results[name]["mean_nvars"]:.1f}')
```

```python
# Summary table
rows = []
for name, res in results.items():
    rows.append({
        'Model': name,
        'Mean score': f"{res['mean_score']:.0f}",
        '± std':      f"{res['std_score']:.0f}",
        'Mean N':     f"{res['mean_N']:.0f}",
        'Mean n_vars':f"{res['mean_nvars']:.1f}",
        'Mean thr':   f"{res['mean_thr']:.3f}",
        'Beats baseline': '✓' if res['mean_score'] > BASELINE else '✗'
    })
df_res = pd.DataFrame(rows).set_index('Model')
df_res = df_res.sort_values('Mean score', ascending=False)
print(f'\nBaseline to beat: {BASELINE}')
display(df_res)
```

```python
# Grouped bar chart: mean ± std per model
names  = list(results.keys())
means  = [results[n]['mean_score'] for n in names]
stds   = [results[n]['std_score']  for n in names]
colors = ['#59a14f' if m > BASELINE else '#e15759' for m in means]

fig, ax = plt.subplots(figsize=(9, 4.5))
bars = ax.bar(names, means, yerr=stds, capsize=5, color=colors,
              edgecolor='white', linewidth=1.1, error_kw={'linewidth':1.5})
ax.axhline(BASELINE, color='#4e79a7', linestyle='--', linewidth=1.8,
           label=f'Baseline = {BASELINE}')
for bar, m, s in zip(bars, means, stds):
    ax.text(bar.get_x() + bar.get_width()/2,
            m + s + 40, f'{m:.0f}', ha='center', va='bottom', fontsize=9)
ax.set_ylabel('Mean CV competition score')
ax.set_title('Competition Score per Model (5-fold CV, optimal N per fold)')
green_p = mpatches.Patch(color='#59a14f', label='Beats baseline')
red_p   = mpatches.Patch(color='#e15759', label='Below baseline')
ax.legend(handles=[green_p, red_p,
                   plt.Line2D([0],[0],color='#4e79a7',linestyle='--',label=f'Baseline {BASELINE}')])
plt.tight_layout()
plt.savefig(FIGURES / '09_model_scores.png')
plt.show()
```

## 4. N_contacted Distribution Across Folds

A stable optimal N across folds indicates the model consistently identifies the same number of high-value customers. High variance in N suggests the model is sensitive to the particular training sample — a risk for generalization.

```python
n_data = []
for name, res in results.items():
    for fold in res['folds']:
        n_data.append({'Model': name, 'N contacted': fold['N'],
                       'Threshold': fold['threshold'], 'n_vars': fold['n_vars']})
df_folds = pd.DataFrame(n_data)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

# N contacted per fold
model_order = sorted(results.keys(), key=lambda n: -results[n]['mean_score'])
palette = {n: '#59a14f' if results[n]['mean_score'] > BASELINE else '#e15759'
           for n in model_order}
sns.boxplot(data=df_folds, x='Model', y='N contacted', order=model_order,
            palette=palette, ax=axes[0], linewidth=1)
axes[0].axhline(1000, color='gray', linestyle=':', linewidth=1, label='Cap = 1000')
axes[0].set_title('Optimal N per Fold (stability)')
axes[0].legend()

# Threshold per fold
sns.violinplot(data=df_folds, x='Model', y='Threshold', order=model_order,
               palette=palette, ax=axes[1], linewidth=1, inner='box')
axes[1].axhline(1/3, color='#f28e2b', linestyle='--', linewidth=1.5, label='1/3 break-even')
axes[1].set_title('Optimal Threshold per Fold (stability)')
axes[1].legend()

plt.tight_layout()
plt.savefig(FIGURES / '10_fold_stability.png')
plt.show()
```

## 5. Per-Fold Score Breakdown

Seeing the raw per-fold scores for each model shows not just the mean but also whether good performance is consistent or driven by one lucky fold.

```python
fold_score_data = []
for name, res in results.items():
    for i, fold in enumerate(res['folds']):
        fold_score_data.append({'Model': name, 'Fold': i+1, 'Score': fold['score']})
df_fs = pd.DataFrame(fold_score_data)

fig, ax = plt.subplots(figsize=(10, 4.5))
sns.stripplot(data=df_fs, x='Model', y='Score', order=model_order,
              palette=palette, jitter=True, size=8, ax=ax, alpha=0.85)
sns.pointplot(data=df_fs, x='Model', y='Score', order=model_order,
              color='black', markers='D', linestyle='none', ax=ax,
              capsize=0.15, errwidth=2, scale=0.7)
ax.axhline(BASELINE, color='#4e79a7', linestyle='--', linewidth=1.8,
           label=f'Baseline = {BASELINE}')
ax.set_ylabel('Competition score (single fold)')
ax.set_title('Per-Fold Scores by Model (dots = individual folds, diamond = mean)')
ax.legend()
plt.tight_layout()
plt.savefig(FIGURES / '11_per_fold_scores.png')
plt.show()
```

## 6. NoVariables Cost Analysis

The NoVariables cost can dominate for high-capacity models that use many features. This plot shows mean n_vars vs mean score — the ideal model is in the top-left: high score, few features.

```python
fig, ax = plt.subplots(figsize=(7, 5))
for name, res in results.items():
    color = '#59a14f' if res['mean_score'] > BASELINE else '#e15759'
    ax.scatter(res['mean_nvars'], res['mean_score'],
               s=120, color=color, zorder=5, edgecolors='white', linewidth=1)
    ax.annotate(name,
                xy=(res['mean_nvars'], res['mean_score']),
                xytext=(5, 5), textcoords='offset points', fontsize=10)

ax.axhline(BASELINE, color='#4e79a7', linestyle='--', linewidth=1.5,
           label=f'Baseline = {BASELINE}')
ax.set_xlabel('Mean n_vars used (NoVariables cost = n_vars × 200)')
ax.set_ylabel('Mean CV competition score')
ax.set_title('Score vs Feature Cost Trade-off')
ax.legend()
plt.tight_layout()
plt.savefig(FIGURES / '12_score_vs_nvars.png')
plt.show()
```

## 7. Model Selection Rationale

Use the cell below to interpret the results and decide which model to take forward.

```python
print('=== FINAL RANKINGS ===')
print(f'Baseline: {BASELINE}\n')
ranked = sorted(results.items(), key=lambda x: -x[1]['mean_score'])
for rank, (name, res) in enumerate(ranked, 1):
    beat = '✓ BEATS BASELINE' if res['mean_score'] > BASELINE else '✗ below baseline'
    gain = res['mean_score'] - BASELINE
    print(f"#{rank} {name:8s}  score={res['mean_score']:6.0f} ± {res['std_score']:4.0f}  "
          f"N={res['mean_N']:5.0f}  n_vars={res['mean_nvars']:5.1f}  "
          f"thr={res['mean_thr']:.3f}  gain={gain:+.0f}  {beat}")
```

### Interpretation

- Any model with **mean CV score > 3 885** beats the current best and should be considered for final submission.
- Prefer models with **lower n_vars** when two models have similar scores — the EUR 200/variable cost compounds.
- **Consistent N across folds** (low std in box plot above) indicates the model reliably finds the same market segment.
- **Threshold near 1/3**: if the optimal threshold is much higher than 1/3, the model's probabilities are poorly calibrated or the positives are very concentrated at high probability — both are fine as long as the CV score is high.
- **LR-L2** will likely score poorly due to using all 500 features (cost = 100 000); it serves as a ceiling on predictive accuracy unbounded by feature cost.
- Tree models (RF, LGBM, XGB) may use many features implicitly — check the n_vars column carefully.

**Next steps**: tune hyperparameters of the winning model(s), consider feature pre-filtering to reduce NoVariables cost before fitting tree models, and generate final test predictions.
